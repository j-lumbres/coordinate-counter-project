```python
import csv
from coordinate_counter import CoordinateCounter
```

```python
coordinates_file = "coordinates.csv"

grid_counter = CoordinateCounter(width=320, height=320)

with open(coordinates_file, mode='r') as file:
    csv_reader = csv.DictReader(file)

    for row in csv_reader:
        xvalue = int(row['x'])
        yvalue = int(row['y'])
        grid_counter.update(xvalue, yvalue)

grid_counter.update(xvalue, yvalue)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 10))
plt.imshow(grid_counter.grid, cmap='hot')
plt.colorbar(label='Number of Hits')
plt.title('Coordinate Distribution Map')
plt.xlabel('X (columns)')
plt.ylabel('Y (rows)')

plt.show()
```
