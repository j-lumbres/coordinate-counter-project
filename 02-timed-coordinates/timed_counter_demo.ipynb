```python
import pandas as pd
from coordinate_timed_counter import CoordinateTimedCounter

df = pd.read_csv("coordinates-with-time.csv.gz")
print(df.head())
```

```python
WINDOW_US = 50_000
counter = CoordinateTimedCounter(width=320, height=320)
frames = []

window_start = df['t'].iloc[0]

for index, row in df.iterrows():
    x = int(row['x'])
    y = int(row['y'])
    t = int(row['t'])

    if t - window_start >= WINDOW_US:
        frames.append(counter.export())
        counter.reset()
        window_start = t
    counter.update(x=x, y=y, t=t)
frames.append(counter.export())

print(f"{len(frames)} frames generated.")
```
